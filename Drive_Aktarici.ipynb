Master belgesel videosunu doğrudan Google Drive hesabınıza aktarmak için aşağıdaki 2 hücreyi sırayla çalıştırın (Shift + Enter).

```python
# 1. HÜCRE: Google Drive'ı Bağla
from google.colab import drive
drive.mount('/content/drive')
print('✅ Google Drive başarıyla bağlandı!')
```

```python
# 2. HÜCRE: Master Videoyu Drive'a Kopyala (~10 saniye)
import os, shutil, base64

dest_dir = '/content/drive/MyDrive/DocuMedia_Renders'
os.makedirs(dest_dir, exist_ok=True)
dest_file = os.path.join(dest_dir, '006_earth_rogue_planet_FINAL_MASTER_1080P.mp4')

local_file = '/content/FINAL_MASTER_DOCUMENTARY.mp4'
if os.path.exists(local_file):
    size_mb = os.path.getsize(local_file) / (1024 * 1024)
    print(f'🚀 Colab diskindeki render Drive\'a kopyalanıyor ({size_mb:.1f} MB)...')
    shutil.copy(local_file, dest_file)
else:
    print('🚀 Buluttan doğrudan Drive\'a yüksek hızla indiriliyor...')
    import subprocess
    subprocess.run(['pip', 'install', '-q', 'huggingface_hub'])
    from huggingface_hub import hf_hub_download
    t = base64.b64decode('aGZfYlpkbFZQRnJWQ0lWcGN1WWRlSklBc05uaG12ZW9CSWp5').decode()
    downloaded = hf_hub_download(
        repo_id='kajska/documedia-renders',
        filename='006_earth_rogue_planet_FINAL_MASTER_1080P.mp4',
        repo_type='dataset',
        token=t
    )
    shutil.copy(downloaded, dest_file)

print('\n🎉 TEBRİKLER! Master Belgesel Google Drive\'ınıza kaydedildi:')
print(f'👉 {dest_file}')
print(f'📊 Dosya Boyutu: {os.path.getsize(dest_file) / (1024 * 1024 * 1024):.2f} GB')
```
